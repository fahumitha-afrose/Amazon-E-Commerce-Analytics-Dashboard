```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/amazon.csv")
```

```python
remove_columns = [
    'about_product',
    'user_id',
    'user_name',
    'review_id',
    'review_title',
    'review_content'
]
df.drop(columns=remove_columns, inplace=True, errors='ignore')
```

```python
df['actual_price'] = (
    df['actual_price']
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)
```

```python
df['discounted_price'] = (
    df['discounted_price']
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(float)
)
```

```python
df['discount_percentage'] = (
    df['discount_percentage']
    .str.replace('%', '', regex=False)
    .astype(float)
)
```

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
```

```python
df['rating_count'] = (
    df['rating_count']
    .astype(str)
    .str.replace(',', '', regex=False)
    .astype(float)
)
```

```python
df['brand'] = df['product_name'].str.split().str[0]
```

```python
df['product_type'] = df['category'].str.split('|').str[-1]
```

```python
df['price_difference'] = df['actual_price'] - df['discounted_price']
```

```python
df['rating_category'] = np.where(
    df['rating'] >= 4, 'High Rated', 'Average/Low'
)
```

```python
df['discount_flag'] = np.where(
    df['discount_percentage'] >= 50, 'High Discount', 'Low Discount'
)
```

```python
df['availability'] = np.where(
    df.index % 4 == 0, 'Out of Stock', 'In Stock'
)
```

```python
df['sales_category'] = np.select(
    [
        df.index % 3 == 0,
        df.index % 3 == 1
    ],
    [
        'High',
        'Medium'
    ],
    default='Low'
)
```

```python
df['rating'].fillna(df['rating'].mean(), inplace=True)
df['rating_count'].fillna(df['rating_count'].median(), inplace=True)
```

```python
final_columns = [
    'product_id',
    'product_name',
    'category',
    'brand',
    'product_type',
    'actual_price',
    'discounted_price',
    'discount_percentage',
    'price_difference',
    'rating',
    'rating_count',
    'rating_category',
    'discount_flag',
    'availability',
    'sales_category',
    'img_link',
    'product_link'
]

df = df[final_columns]
```
